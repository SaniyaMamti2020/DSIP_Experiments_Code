```python
import cv2
import numpy as np
import matplotlib.pyplot as plt
```

```python
image = cv2.imread('img1.png')

if image is None:
    print("Error: Could not read the image.")
```

```python
# Display in BGR format (OpenCV default)
plt.figure(figsize=(6, 6))
plt.imshow(image)
plt.title("Input Image")
plt.axis("Off")
plt.show()
```

```python
# Apply BGR to RGB conversion for correct color display in matplotlib
image2 = cv2.cvtColor(image, cv2.COLOR_BGR2RGB)
plt.imshow(image2)
plt.title("Input Image")
plt.axis("Off")
plt.show()
```

```python
# Apply Grayscale Transformation in way 1
gray_image = cv2.cvtColor(image, cv2.COLOR_BGR2GRAY)
plt.imshow(gray_image, cmap='gray')
plt.title("Grayscale Image")
plt.axis("Off")
plt.show()
```

```python
image = cv2.imread('img1.png',cv2.IMREAD_GRAYSCALE)

if image is None:
    print("Error: Could not read the image.")
```

```python
# Apply Grayscale Transformation in way 2
plt.imshow(image, cmap='gray')
plt.title("Grayscale Image")
plt.axis("Off")
plt.show()
```

```python
# Apply Negative Transformation
negative_image = 255 - image
plt.figure(figsize=(6, 6))
plt.imshow(negative_image, cmap='gray')
plt.title("Negative Image")
plt.axis("Off")
plt.show()
```

```python
c = 255 / np.log(1 + np.max(image))
print(c)
```

```python
# Apply Log Transformation
c = 50
log_image = c * np.log(1 + image.astype(np.float32))
log_image = np.array(log_image, dtype=np.uint8)

plt.figure(figsize=(6, 6))
plt.imshow(log_image, cmap='gray')
plt.title("Log Transformed Image")
plt.axis("Off")
plt.show()
```

```python
# Apply Power Law gamma correction
gamma = 0.5
gamma_image = np.array(255 * (image / 255) ** gamma,dtype=np.uint8)

plt.figure(figsize=(6, 6))
plt.imshow(gamma_image, cmap='gray')
plt.title("Gamma Transformed Image")
plt.axis("Off")
plt.show()
```

```python
threshold_value = 125
threshold_value, threshold_image = cv2.threshold(image, threshold_value, 255, cv2.THRESH_BINARY)
plt.figure(figsize=(6, 6))
plt.title("Thresholded Image")
plt.imshow(threshold_image, cmap='gray')
plt.axis("Off")
plt.show()
```
