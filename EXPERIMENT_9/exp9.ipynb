```python
import cv2
import numpy as np
import matplotlib.pyplot as plt
```

```python
image = cv2.imread("ss.jpg")
plt.imshow(cv2.cvtColor(image, cv2.COLOR_BGR2RGB))
plt.axis("off")
plt.show()
```

```python
mean_3x3 = cv2.blur(image, (5,5))

plt.figure(figsize=(12,4))

plt.subplot(1,2,1)
plt.imshow(image)
plt.title('Original Image')
plt.axis("off")

plt.subplot(1,2,2)
plt.imshow(mean_3x3)
plt.title('Blurred Image')
plt.axis("off")
```

```python
weighted_kernel = np.array([
    [1,1,1],
    [1,2,1],
    [1,1,1]], dtype = np.float32)

weighted_kernel = weighted_kernel / 10

weighted_mean = cv2.filter2D(image,-1,weighted_kernel)

plt.figure(figsize=(8,6))
plt.subplot(1,2,1)
plt.imshow(image)
plt.title('Original Image')
plt.axis("off")

plt.subplot(1,2,2)
plt.imshow(weighted_mean)
plt.title('Weighted Image')
plt.axis("off")
```

```python
g0 = cv2.GaussianBlur(image,(5,5),sigmaX=0)
g1 = cv2.GaussianBlur(image,(5,5),sigmaX=1)
g2 = cv2.GaussianBlur(image,(5,5),sigmaX=2)

plt.figure(figsize=(12,4))
plt.subplot(1,5,1)
plt.imshow(image)
plt.title('Original Image')
plt.axis("off")

plt.subplot(1,5,2)
plt.imshow(cv2.cvtColor(g0, cv2.COLOR_BGR2RGB))
plt.title('Gaussian Blur 0')
plt.axis("off")

plt.subplot(1,5,3)
plt.imshow(cv2.cvtColor(g1, cv2.COLOR_BGR2RGB))
plt.title('Gaussian Blur 1')
plt.axis("off")

plt.subplot(1,5,4)
plt.imshow(cv2.cvtColor(g2, cv2.COLOR_BGR2RGB))
plt.title('Gaussian Blur 2')
plt.axis("off")

plt.show()
```

```python
laplacian = cv2.Laplacian(mean_3x3, cv2.CV_64F)
laplacian_abs = cv2.convertScaleAbs(laplacian)

plt.figure(figsize=(8,6))
plt.subplot(1,2,1)
plt.imshow(image)
plt.title('Original Image')
plt.axis("off")

plt.subplot(1,2,2)
plt.imshow(laplacian_abs, cmap='gray')
plt.title('Laplacian Filter Image')
plt.axis("off")
```
